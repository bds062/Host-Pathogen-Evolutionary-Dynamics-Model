```python
param1_values = np.linspace(0, 365, 15) #vertical axis
param2_values = np.linspace(0, 365, 15) #horizontal axis



# X_0 must seed BOTH genotypes' infected classes (indices 3-5 and 7-9). With genotype 2
# (Ij2/Iv2/If2) starting at 0, dIj2 = Bj*Sj*If2 - ... etc. are all proportional to the
# already-zero genotype-2 populations, so it can never establish and every ESS sweep
# point degenerates to "genotype 1 always wins" regardless of germination/maturity.
# Seed symmetrically, matching Fig 4.
model = Model(X_0 = (5, 0, 0, 2.5, 0, 0, 0, 2.5, 0, 0), t = (0, 365 * 400), germination_time = 212.5, Bj = 0.2 / 365, infected_time=182.5, infected_time2=182.5)
# graph_model = GraphModel(model)
raster_model = RasterModel(model, 'infected_time', 'infected_time2', param1_values, param2_values)

# print("Maturation rate:", model.maturity*365)
print("BetaJ: ", model.Bj*365)
print("BetaA: ", model.Bv*365)


start_time = time.time()
raster_model.plot_fig3(
    germination_values=np.linspace(62.5, 302.5, 9),
    flowering_values=np.linspace(0, 365, 40),
    base_maturity=4/365,
    maturity_multipliers=(4, 8, 16, 32),
    save_path="../figs/fig3.pdf",
    results_save_path="../figs/fig3_embedding.npz",
)
end_time = time.time()
print(f"\nTotal execution time: {end_time - start_time:.2f} seconds")
```

```python
model = Model(X_0 = (5, 0, 0, 2.5, 0, 0, 0, 2.5, 0, 0), t = (0, 365 * 400))
raster = RasterModel(model, 'infected_time', 'infected_time2', None, None)

germination_values = np.linspace(62.5, 302.5, 18)   # x-axis points (18 values)
flowering_values   = np.linspace(0, 365,40)        # raster grid (40×40 per point)

model_configs = [
    {},   # basic — uses self.model defaults
        # self.Bj = 0.2 / 365
        # self.Bv = 0.025 / 365
        # self.Bf = 0.025 / 365
        # self.births = 4 / 365
        # self.gamma = 0.0001
        # self.death = .5 / 365
        # self.maturity = 4 / 365
        # self.g = 2 / 365
    {     # Silene latifolia (Slowinski et al. 2025)
        'Bj': 0.6/365, 'Bv': 0.4/365, 'Bf': 0.4/365,
        'births': 8/365, 'gamma': 0.001, 'death': 0.36/365,
        'maturity': 0.85/365, 'g': 2/365,
        # self.Bj = .6/365 #(0.93+0.27)/2
        # self.Bv = 0.4/365
        # self.Bf = 0.4/365
        # self.births = 8/365 #paper says 2/365
        # self.gamma = 0.001
        # self.death = .36/365
        # self.maturity = .85/365 #(.7+.9)/2
        # self.g = 2/365
    },
    {     # Dianthus pavonius (Bruns et al. 2017)
        'Bj': 3.38e-4, 'Bv': 3.38e-5, 'Bf': 3.38e-5,
        'births': 1.3/365, 'gamma': 1.5e-4, 'death': 0.187/365,
        'maturity': 8/365, 'g': 2/365,
        # self.Bj = 3.38e-4      # juvenile density-dependent transmission: s * beta_d
        # self.Bv = 3.38e-5      # vegetative adult density-dependent transmission: beta_d
        # self.Bf = 3.38e-5      # flowering adult density-dependent transmission: beta_d
        # self.births = 1.3/365      # b: establishment / new juvenile production from healthy flowering plants #DAY SCALED
        # self.gamma = 1.5e-4    # k: density-dependent competition effect
        # self.death = 0.187/365     # average of mu_f = 0.116 and mu_v = 0.258 #DAY SCALED
        # self.maturity = 8/365    # all surviving juveniles leave Sj and enter Sv in your model #DAY SCALED, new default
        # self.g = 2/365           # seed-to-juvenile germination/establishment over the annual step #DAY SCALED
    },
]

results = raster.plot_fig4(
    germination_values = germination_values,
    flowering_values   = flowering_values,
    model_configs      = model_configs,
    config_labels      = ['Basic', 'Silene latifolia', 'Dianthus pavonius'],
    results_save_path  = '../figs/fig4.6_embedding.npz',  # reload later with np.load(...); already resumes from cache automatically
    save_path          = '../figs/fig4.pdf',
)
```
